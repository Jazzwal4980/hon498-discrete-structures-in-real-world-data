# Degrees

## What is a degree?
The degree of a vertex in a graph is the number of edges incident to it. The word incident in the graph theory context describes the connection between a node (vertex) and an edge (the connecting line). Below is an example of a graph with its degrees notated. As you can see, the vertex A has a degree of 2 because it has 2 connections to other edges. Similarly, vertex E has a degree of 4 because it has 4 connections to other edges. 

```python
import matplotlib.pyplot as plt
import networkx as nx

# 1. Create an undirected graph
G = nx.Graph()

# 2. Add edges to match the example described in your text
# Node 'A' is the far-left node (Degree = 2)
# Node '3' is the central node (Degree = 4)
edges = [
    ("A", "E"),
    ("A", "B"),
    ("E", "B"),
    ("E", "C"),
    ("E", "D"),
]
G.add_edges_from(edges)

# 3. Calculate degrees for every node
degrees = dict(G.degree())

# 4. Set fixed positions for a clean layout (Far-left node on the left, node '3' centered)
pos = {
    "A": (-2, 0),  # Far-left node (Degree 2)
    "E": (0, 0),  # Central node (Degree 4)
    "B": (-1, 1),  # Top node connecting A and 3
    "C": (1.5, 1),  # Top-right neighbor
    "D": (1.5, -1),  # Bottom-right neighbor
}

# 5. Create labels showing both the node name/value and its degree
node_labels = {node: f"{node}\ndeg({deg})" for node, deg in degrees.items()}

# 6. Plot the graph
plt.figure(figsize=(9, 5))

# Draw nodes
nx.draw_networkx_nodes(
    G, pos, node_size=2500, node_color="#4B9CD3", edgecolors="black", linewidths=1.5
)

# Draw edges
nx.draw_networkx_edges(G, pos, width=2, edge_color="#555555")

# Draw custom labels inside the nodes
nx.draw_networkx_labels(
    G, pos, labels=node_labels, font_size=9, font_color="white", font_weight="bold"
)

# Customize title and display grid settings
plt.title(
    "Undirected Graph: Node Degree Example",
    fontsize=14,
    fontweight="bold",
    pad=15,
)
plt.axis("off")  # Turn off standard x-y coordinate axes
plt.tight_layout()
plt.show()
```

## How is a degree notated
In the case of a simple graph, a degree can be formally written as deg(u), where u is a vertex of a graph. Using our previous graph and vertices, vertex A can be denoted as deg(2), and vertex E can be denoted as deg(4). Directed graphs have a similar notation but vary slightly due to the direction of edges, which can further be classified as in-degree and out-degree. The in-degree of a vertex is the number of edges where the edge is pointed towards the vertex, also known as a terminal vertex. The out-degree of a vertex is the number of edges where the edge is pointed away from the vertex, also known as the initial vertex. Directed graphs have comma-paired values to represent first the in-degrees and then the out-degrees. Equation-wise, in-degree is often denoted as deg-(u) and out-degree is often denoted as deg+(u), where u is the vertex. Let us look at an example down below.

```python
import matplotlib.pyplot as plt
import networkx as nx

# 1. Create a directed graph (DiGraph)
G = nx.DiGraph()

# 2. Add 8 vertices labelled A through H
nodes = ["A", "B", "C", "D", "E", "F", "G", "H"]
G.add_nodes_from(nodes)

# 3. Add directed edges to create a balanced 8-vertex network
edges = [
    ("A", "B"),
    ("A", "C"),
    ("B", "D"),
    ("B", "E"),
    ("C", "E"),
    ("C", "F"),
    ("D", "G"),
    ("E", "D"),
    ("E", "G"),
    ("E", "H"),
    ("F", "E"),
    ("F", "H"),
    ("G", "H"),
]
G.add_edges_from(edges)

# 4. Calculate in-degree (deg-) and out-degree (deg+) for every node
in_degrees = dict(G.in_degree())
out_degrees = dict(G.out_degree())

# 5. Set positions for an 8-node layout across 4 columns
pos = {
    "A": (-3.0, 0.0),  # Column 1 (Source)
    "B": (-1.0, 1.2),  # Column 2 (Top)
    "C": (-1.0, -1.2),  # Column 2 (Bottom)
    "D": (1.0, 1.2),  # Column 3 (Top)
    "E": (0.0, 0.0),  # Column 3 (Center hub)
    "F": (1.0, -1.2),  # Column 3 (Bottom)
    "G": (2.5, 0.8),  # Column 4 (Top-Right)
    "H": (2.5, -0.8),  # Column 4 (Bottom-Right)
}

plt.figure(figsize=(11, 6))

# 6. Draw blue node circles
nx.draw_networkx_nodes(
    G, pos, node_size=1800, node_color="blue", edgecolors="black", linewidths=1.0
)

# 7. Draw straight, crisp directed arrows
nx.draw_networkx_edges(
    G,
    pos,
    width=1.5,
    edge_color="#333333",
    arrowstyle="-|>",
    arrowsize=18,
    node_size=1800,
    connectionstyle="arc3,rad=0.0",  # Straight lines
)

# 8. Draw letter labels inside the nodes
node_labels = {node: node for node in G.nodes()}
nx.draw_networkx_labels(
    G, pos, labels=node_labels, font_size=14, font_color="white", font_weight="bold"
)

# 9. Add (In, Out) degree notation above each node (Red for In, Blue for Out)
for node, (x, y) in pos.items():
    in_deg = in_degrees[node]
    out_deg = out_degrees[node]

    # Vertical position offset above node
    y_offset = 0.38

    # Red for In-degree
    plt.text(
        x,
        y + y_offset,
        f"{in_deg}",
        color="red",
        fontsize=13,
        fontweight="bold",
        ha="right",
        va="center",
    )
    # Blue for Out-degree
    plt.text(
        x,
        y + y_offset,
        f", {out_deg}",
        color="blue",
        fontsize=13,
        fontweight="bold",
        ha="left",
        va="center",
    )

# Formatting adjustments
plt.xlim(-3.8, 3.3)
plt.ylim(-2.0, 2.0)
plt.title(
    "8-Vertex Directed Graph: In-Degree (Red) & Out-Degree (Blue)",
    fontsize=14,
    fontweight="bold",
    pad=20,
)
plt.axis("off")
plt.tight_layout()
plt.show()
```

In the example above, we can see above that vertex B has 3 total connections: 1 in-degree (denoted in red) and 2 out-degree (denoted in blue). Vertex B would be denoted as deg-(1) and deg+(2). Another example: vertex H has 3 total connections: 3 in-degrees (denoted in red) and 0 out-degrees (denoted in blue). Vertex H is denoted as deg-(3) and deg+(0).

## Handshaking Theoreom
Generally, degree values of all types of graphs can range from 0 to |V| - 1, where |V| is the number of vertices within the graph. If a vertex has degree 0, then we say that the vertex is isolated from the rest of the graph. Additionally, if the vertex has degree 1, then we say that the vertex is a pendant. If we expand upon this and make a note of all the adjacent vertices, the formal definition is called a neighborhood. If we want to quickly calculate the total number of degrees of a simple graph, we can use what’s called the Handshaking Theorem. The Handshaking Theorem states that since each edge contributes twice (one for initial and one for the terminal vertex), the sum of degrees is equal to twice the number of edges. The Handshaking Theorem also has an interesting result: the number of vertices with odd degree is even. Let us observe one last property with the Handshaking Theorem for directed graphs: then we can say that the sum of deg-(u) is equal to the sum of deg+(u), which is ultimately equal to the number of edges in the graph. Let us take a look at an example where we can see all the properties previously mentioned and the Handshaking Theorem.  

```python
import matplotlib.pyplot as plt
import networkx as nx

# 1. Create an undirected graph
G = nx.Graph()

# 2. Add 16 vertices labeled A through P
nodes = [chr(i) for i in range(ord("A"), ord("P") + 1)]
G.add_nodes_from(nodes)

# 3. Add edges to create isolated nodes (A, B), a pendant node (C), and connected clusters
edges = [
    # Pendant connection (Degree 1)
    ("C", "D"),
    # Core cluster
    ("D", "E"),
    ("D", "F"),
    ("E", "F"),
    ("E", "G"),
    ("F", "G"),
    # Middle network
    ("G", "H"),
    ("H", "I"),
    ("H", "J"),
    ("I", "J"),
    ("I", "K"),
    ("J", "L"),
    # Right cluster
    ("K", "M"),
    ("K", "N"),
    ("L", "N"),
    ("M", "N"),
    ("N", "O"),
    ("O", "P"),
]
G.add_edges_from(edges)

# 4. Calculate degrees for every node
degrees = dict(G.degree())
total_degrees = sum(degrees.values())
num_edges = G.number_of_edges()

# Count odd and even degree vertices
odd_degree_nodes = [node for node, deg in degrees.items() if deg % 2 != 0]
even_degree_nodes = [node for node, deg in degrees.items() if deg % 2 == 0]

# 5. Fixed positions across 16 nodes for a clean, readable layout
pos = {
    "A": (-3.0, 1.2),  # Isolated
    "B": (-3.0, -1.2),  # Isolated
    "C": (-2.0, 0.0),  # Pendant (Degree 1)
    "D": (-1.2, 0.0),
    "E": (-0.6, 0.8),
    "F": (-0.6, -0.8),
    "G": (0.0, 0.0),
    "H": (0.6, 0.0),
    "I": (1.2, 0.8),
    "J": (1.2, -0.8),
    "K": (2.0, 0.8),
    "L": (2.0, -0.8),
    "M": (2.8, 1.2),
    "N": (2.8, 0.0),
    "O": (3.4, 0.0),
    "P": (4.0, 0.0),
}

plt.figure(figsize=(12, 6.5))

# 6. Draw nodes: Color odd-degree nodes light red and even-degree nodes light blue
nx.draw_networkx_nodes(
    G,
    pos,
    nodelist=even_degree_nodes,
    node_size=1200,
    node_color="#87CEEB",
    edgecolors="black",
    linewidths=1.2,
    label="Even Degree Node",
)
nx.draw_networkx_nodes(
    G,
    pos,
    nodelist=odd_degree_nodes,
    node_size=1200,
    node_color="#FFA07A",
    edgecolors="black",
    linewidths=1.2,
    label="Odd Degree Node",
)

# 7. Draw undirected edges
nx.draw_networkx_edges(G, pos, width=1.5, edge_color="#444444")

# 8. Draw labels showing Node Name and Degree: Node\ndeg(v)
node_labels = {node: f"{node}\ndeg: {deg}" for node, deg in degrees.items()}
nx.draw_networkx_labels(
    G, pos, labels=node_labels, font_size=8, font_color="black", font_weight="bold"
)

plt.title(
    "16-Vertex Simple Undirected Graph: Handshaking Theorem Demonstration",
    fontsize=13,
    fontweight="bold",
    pad=15,
)

# 9. Clean, spaced-out legend configuration
plt.legend(
    loc="upper right",
    framealpha=0.9,
    fontsize=10,
    markerscale=0.4,  # Scales down circle symbols in the legend
    labelspacing=1.2,  # Adds vertical space between legend rows
    handletextpad=1.0,  # Adds horizontal space between circle and text
    borderpad=1.0,  # Adds padding around the inner border of legend
)

# Adjust plot boundaries to keep top and bottom clean
plt.ylim(-2.2, 2.2)
plt.axis("off")
plt.tight_layout()
plt.subplots_adjust(bottom=0.22, top=0.92)
plt.show()
```

In the example above, there are a total of 16 vertices, 18 edges, and 36 degrees. we can see that both vertices A and B are isolated since deg(0). We can also see that vertices C and P are pendants since deg(1). According to the Handshake Theorem which we previously defined:
$$
\underbrace{\sum_{v \in V} \text{deg}(v)}_{\text{Sum of all vertex degrees}} = \underbrace{2|E|}_{\text{Twice the number of edges}}
$$

Substituting the values into the equation, we can see that the Handshake Theorem holds true:
$$
\sum_{v \in V} \text{deg}(v) = {2|18|} \rightarrow 36 = 36
$$


# Paths and Connectivity

## Paths
What if instead of knowing how many edges are connected to a vertex, we want to know if a route exists from vertex v0 to vi? First, there is a base definition that we must define where we do not care if the vertices have to be distinct, in which we call it a Walk. Further extending the idea, a closed walk is called a circuit (v0 = vi). When the length from vo to vi is a consecutive sequence of distinct vertices, it is known as a path. Symbolically, it’s represented as: {vo, v1, … , vi - 1, vi}, where vi and vi+1 are adjacent. Simply put, a path is a walk with distinct vertices. We refer to this type of path as a simple path since the same vertex does not appear twice. Further extending the idea, a closed path is called a cycle since there are no repeated vertices and a cycle graph is one where all the degrees are equal.

## Connectivity
Now that we have established the definition of a path, we can also define connectivity. If there exists a path from vertex u to vertex v, then we can confirm that v is reachable from u (u → v), which we can conclude that u is connected to v. Expanded broadly to the whole graph, we can conclude that a whole graph is connected if there exists an x-y path for all vertices within the graph. We can denote the number of connected graph or components with the notation K(G), where K(G) represents the number of connected components. Let us look at the example below to see both cycles and connectivity.

```python
import matplotlib.pyplot as plt
import networkx as nx

# 1. Create an undirected graph
G = nx.Graph()

# 2. Add 16 vertices labeled A through P
nodes = [chr(i) for i in range(ord("A"), ord("P") + 1)]
G.add_nodes_from(nodes)

# 3. Add edges to form 3 distinct connected components (Discontinuous Graph)
edges = [
    # Component 1 (Vertices A, B, C, D, E)
    ("A", "B"),
    ("B", "D"),
    ("D", "E"),
    ("E", "C"),
    ("C", "A"),
    # Component 2 (Vertices F, G, H, I, J)
    ("F", "G"),
    ("G", "H"),
    ("H", "I"),
    ("I", "J"),
    ("G", "I"),
    # Component 3 (Vertices K, L, M, N, O, P)
    ("K", "L"),
    ("L", "M"),
    ("M", "N"),
    ("N", "O"),
    ("O", "P"),
    ("M", "P"),
    ("K", "N"),
]
G.add_edges_from(edges)

# 4. Extract connected components
components = list(nx.connected_components(G))
num_components = len(components)

# Define distinct colors for the 3 components
component_colors = ["#87CEEB", "#FFA07A", "#98FB98"]  # Sky Blue, Light Salmon, Pale Green

# 5. Fixed positions to visually separate the 3 components across the layout
pos = {
    # Component 1 (Left Cluster)
    "A": (-3.2, 0.8),
    "B": (-3.2, -0.8),
    "C": (-2.2, 0.8),
    "D": (-2.2, -0.8),
    "E": (-1.4, 0.0),
    # Component 2 (Center Cluster)
    "F": (-0.4, 1.0),
    "G": (-0.4, -0.2),
    "H": (0.4, 1.0),
    "I": (0.4, -0.2),
    "J": (0.0, -1.0),
    # Component 3 (Right Cluster)
    "K": (1.4, 0.8),
    "L": (1.4, -0.8),
    "M": (2.2, 0.8),
    "N": (2.2, -0.8),
    "O": (3.0, 0.8),
    "P": (3.0, -0.8),
}

plt.figure(figsize=(12, 6.5))

# 6. Draw each component with a unique color and label for the legend
for idx, comp in enumerate(components):
    comp_nodes = list(comp)
    nx.draw_networkx_nodes(
        G,
        pos,
        nodelist=comp_nodes,
        node_size=1200,
        node_color=component_colors[idx],
        edgecolors="black",
        linewidths=1.2,
        label=f"Component {idx + 1} ({len(comp_nodes)} Vertices)",
    )

# 7. Draw straight, clean edges
nx.draw_networkx_edges(G, pos, width=1.5, edge_color="#444444")

# 8. Draw node labels (Letter names) inside the nodes
node_labels = {node: node for node in G.nodes()}
nx.draw_networkx_labels(
    G, pos, labels=node_labels, font_size=10, font_color="black", font_weight="bold"
)

plt.title(
    "16-Vertex Undirected Graph",
    fontsize=13,
    fontweight="bold",
    pad=15,
)

# 9. Clean, spaced-out legend configuration
plt.legend(
    loc="upper right",
    framealpha=0.9,
    fontsize=9.5,
    markerscale=0.4,
    labelspacing=1.2,
    handletextpad=1.0,
    borderpad=1.0,
)

# Layout bounds
plt.xlim(-3.8, 3.6)
plt.ylim(-1.8, 1.8)
plt.axis("off")
plt.tight_layout()
plt.subplots_adjust(bottom=0.22, top=0.92)
plt.show()
```

In the above example, there are 3 total components within this 16-vertex graph, which can be denoted as K(3). Since $K(G) > 1$, we can conclude that the graph overall is disconnected. To imagine a graph that is connected, each of the components on its own is connected, which we will analyze individually. 

### Component 1
Analyzing component 1 individually, we can see that it is a connected undirected graph. There is a Walk from $A \rightarrow E$ and also a path since all of the vertices are unique. This is a special type of graph, called a "cycle graph," since for all vertices have the same number of degrees: deg(2).

### Component 2
Analyzing component 2 individually, we can see that it is a connected undirected graph. One possible Walk from $F \rightarrow J$ is $F \rightarrow G \rightarrow H \rightarrow I \rightarrow J$. This Walk can also be considered a Path since all of the vertices are unique and are not repeated. One possible path from $G \rightarrow G$ is $G \rightarrow H \rightarrow I \rightarrow G$, which we can define as a closed path, also known as a cycle.

### Component 3
Analyzing component 3 individually, we can see that it is a connected undirected graph. One possible Walk from $K \rightarrow L$ is $K \rightarrow N \rightarrow M \rightarrow P \rightarrow O \rightarrow N \rightarrow M \rightarrow L$. This would NOT be a path since the vertices are not unique and are repeated. A possible path from $K \rightarrow L$ is $K \rightarrow N \rightarrow O \rightarrow P \rightarrow M \rightarrow L$. 

## Properties of Connectivity

Some important graph relation terms we will define are symmetric, reflexive, and transitive. 

### Symmetric
A graph is symmetric (arc-transitive) if, for any two pairs of adjecent vertices (u, v) and (x,y) are mapped to each other. Specifically, there has to be graph automorphism $f: V(G) \rightarrow V(G)$ such that $f(u) = x$ and $f(v) = y$ simultaneously. The formal mathematical definition is listed below:
$$
\begin{aligned}
\text{A relation on A is symmetic if } \forall x, y \in A, \text{if xRy then yRx} \\ 
\text{\small where xRy means X related to Y and yRx means Y related to X}
\end{aligned}
$$

Symmetric undirected and directed graphs can be seen below:

```python
import matplotlib.pyplot as plt
import networkx as nx

# Create figure with 2 subplots side-by-side
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(12, 5))

# ==========================================
# 1. UNDIRECTED GRAPH (2 Symmetric Vertices)
# ==========================================
G_undirected = nx.Graph()
G_undirected.add_edge("A", "B")

pos_undirected = {"A": (-1.0, 0.0), "B": (1.0, 0.0)}

# Draw Undirected Nodes
nx.draw_networkx_nodes(
    G_undirected,
    pos_undirected,
    ax=ax1,
    node_size=2000,
    node_color="#87CEEB",
    edgecolors="black",
    linewidths=1.5,
)

# Draw Undirected Edge
nx.draw_networkx_edges(
    G_undirected, pos_undirected, ax=ax1, width=2.5, edge_color="#333333"
)

# Draw Labels
nx.draw_networkx_labels(
    G_undirected,
    pos_undirected,
    ax=ax1,
    font_size=14,
    font_color="black",
    font_weight="bold",
)

ax1.set_title(
    "1. Undirected Graph (2 Vertices)\nSymmetric: f(A)=B, f(B)=A",
    fontsize=12,
    fontweight="bold",
    pad=15,
)
ax1.set_xlim(-2.0, 2.0)
ax1.set_ylim(-1.0, 1.0)
ax1.axis("off")


# ==========================================
# 2. DIRECTED GRAPH (2 Symmetric Vertices)
# ==========================================
# To be symmetric, A and B must point to each other symmetrically (A <-> B)
G_directed = nx.DiGraph()
G_directed.add_edges_from([("A", "B"), ("B", "A")])

pos_directed = {"A": (-1.0, 0.0), "B": (1.0, 0.0)}

# Draw Directed Nodes
nx.draw_networkx_nodes(
    G_directed,
    pos_directed,
    ax=ax2,
    node_size=2000,
    node_color="#FFA07A",
    edgecolors="black",
    linewidths=1.5,
)

# Draw Curved Directed Arrows so both directions (A->B and B->A) are clear
nx.draw_networkx_edges(
    G_directed,
    pos_directed,
    ax=ax2,
    width=2.0,
    edge_color="#333333",
    arrowstyle="-|>",
    arrowsize=20,
    node_size=2000,
    connectionstyle="arc3,rad=0.2",  # Curved arcs for bidirectional edges
)

# Draw Labels
nx.draw_networkx_labels(
    G_directed,
    pos_directed,
    ax=ax2,
    font_size=14,
    font_color="black",
    font_weight="bold",
)

ax2.set_title(
    "2. Directed Graph (2 Vertices)\nSymmetric: f(A)=B, f(B)=A (Mutual Edges)",
    fontsize=12,
    fontweight="bold",
    pad=15,
)
ax2.set_xlim(-2.0, 2.0)
ax2.set_ylim(-1.0, 1.0)
ax2.axis("off")

plt.tight_layout()
plt.subplots_adjust(bottom=0.25)
plt.show()
```

### Reflexive
A graph is reflexive if for all the vertices u in the graph, there exists an edge (u, u), which is essentially a self-loop that connects to itself. The formal mathematical definition is listed below:
$$
\begin{aligned}
\text{A relation A is reflexive if: } xRx, \forall x \in A \\
\text{\small where xRx means X related to X}
\end{aligned}
$$

Reflexive undirected and directed graphs can be seen below:

```python
import matplotlib.pyplot as plt
import networkx as nx

# Create figure with 2 subplots side-by-side
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(11, 4.5))

# ==========================================
# 1. UNDIRECTED REFLEXIVE GRAPH (Single Node)
# ==========================================
G_undirected = nx.Graph()
G_undirected.add_node("A")
G_undirected.add_edge("A", "A")  # Self-loop

pos_u = {"A": (0, 0)}

# Draw Undirected Node
nx.draw_networkx_nodes(
    G_undirected,
    pos_u,
    ax=ax1,
    node_size=2500,
    node_color="#87CEEB",
    edgecolors="black",
    linewidths=1.5,
)

# Draw Self-Loop Edge
nx.draw_networkx_edges(
    G_undirected, pos_u, ax=ax1, width=2.0, edge_color="#333333"
)

# Draw Node Label
nx.draw_networkx_labels(
    G_undirected,
    pos_u,
    ax=ax1,
    font_size=14,
    font_color="black",
    font_weight="bold",
)

ax1.set_title(
    "1. Undirected Reflexive Graph\nSingle Vertex with Self-Loop {A, A}",
    fontsize=11,
    fontweight="bold",
    pad=15,
)
ax1.set_xlim(-1.5, 1.5)
ax1.set_ylim(-1.5, 1.5)
ax1.axis("off")


# ==========================================
# 2. DIRECTED REFLEXIVE GRAPH (Single Node)
# ==========================================
G_directed = nx.DiGraph()
G_directed.add_node("A")
G_directed.add_edge("A", "A")  # Directed self-loop (A -> A)

pos_d = {"A": (0, 0)}

# Draw Directed Node
nx.draw_networkx_nodes(
    G_directed,
    pos_d,
    ax=ax2,
    node_size=2500,
    node_color="#FFA07A",
    edgecolors="black",
    linewidths=1.5,
)

# Draw Directed Self-Loop Arrow
nx.draw_networkx_edges(
    G_directed,
    pos_d,
    ax=ax2,
    width=2.0,
    edge_color="#333333",
    arrowstyle="-|>",
    arrowsize=20,
    node_size=2500,
)

# Draw Node Label
nx.draw_networkx_labels(
    G_directed,
    pos_d,
    ax=ax2,
    font_size=14,
    font_color="black",
    font_weight="bold",
)

ax2.set_title(
    "2. Directed Reflexive Graph\nSingle Vertex with Self-Loop (A → A)",
    fontsize=11,
    fontweight="bold",
    pad=15,
)
ax2.set_xlim(-1.5, 1.5)
ax2.set_ylim(-1.5, 1.5)
ax2.axis("off")

plt.tight_layout()
plt.subplots_adjust(bottom=0.28)
plt.show()
```

### Transitive

A graph is said to be transitive if there exists a path from vertices X to Y and from Y to Z, then there also must be an edge from X to Z. The formal mathematical definition is listed below:
$$
\begin{aligned}
\text{A relation on A is transitive if } \forall x, y, z \in A \text{, if xRy and yRz, then xRz}\\
\text{\small where xRy means X related to Y, yRz means Y related to Z, and xRz means X related to Z}
\end{aligned}
$$

Transitive undirected and directed graphs can be seen below:

```python
import matplotlib.pyplot as plt
import networkx as nx

# Create figure with 2 subplots side-by-side
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(12, 5))

# Triangular layout positions for 3 vertices (X, Y, Z)
pos = {
    "X": (-1.0, -0.8),  # Bottom-Left
    "Y": (0.0, 0.8),  # Top-Center
    "Z": (1.0, -0.8),  # Bottom-Right
}

# ==========================================
# 1. UNDIRECTED TRANSITIVE GRAPH (K3 Triangle)
# ==========================================
G_undirected = nx.Graph()
G_undirected.add_nodes_from(["X", "Y", "Z"])

# In undirected graphs, transitivity requires that if {X, Y} and {Y, Z} exist, {X, Z} must also exist.
# This forms a complete 3-clique (triangle graph K_3).
G_undirected.add_edges_from([("X", "Y"), ("Y", "Z"), ("X", "Z")])

# Draw Undirected Nodes
nx.draw_networkx_nodes(
    G_undirected,
    pos,
    ax=ax1,
    node_size=2000,
    node_color="#87CEEB",
    edgecolors="black",
    linewidths=1.5,
)

# Draw Undirected Edges
nx.draw_networkx_edges(
    G_undirected, pos, ax=ax1, width=2.0, edge_color="#333333"
)

# Draw Node Labels
nx.draw_networkx_labels(
    G_undirected,
    pos,
    ax=ax1,
    font_size=14,
    font_color="black",
    font_weight="bold",
)

ax1.set_title(
    "1. Undirected Transitive Graph\n{X, Y} and {Y, Z} ⟹ {X, Z} Exists (K₃)",
    fontsize=11,
    fontweight="bold",
    pad=15,
)
ax1.set_xlim(-1.6, 1.6)
ax1.set_ylim(-1.3, 1.3)
ax1.axis("off")


# ==========================================
# 2. DIRECTED TRANSITIVE GRAPH (TOSET Chain)
# ==========================================
G_directed = nx.DiGraph()
G_directed.add_nodes_from(["X", "Y", "Z"])

# Transitive property: X -> Y and Y -> Z implies shortcut X -> Z
G_directed.add_edges_from(
    [
        ("X", "Y"),  # X -> Y
        ("Y", "Z"),  # Y -> Z
        ("X", "Z"),  # X -> Z (Transitive shortcut edge)
    ]
)

# Draw Directed Nodes
nx.draw_networkx_nodes(
    G_directed,
    pos,
    ax=ax2,
    node_size=2000,
    node_color="#FFA07A",
    edgecolors="black",
    linewidths=1.5,
)

# Draw Directed Arrows (Straight lines for clear hierarchy)
nx.draw_networkx_edges(
    G_directed,
    pos,
    ax=ax2,
    width=2.0,
    edge_color="#333333",
    arrowstyle="-|>",
    arrowsize=20,
    node_size=2000,
    connectionstyle="arc3,rad=0.0",  # Straight lines
)

# Draw Node Labels
nx.draw_networkx_labels(
    G_directed,
    pos,
    ax=ax2,
    font_size=14,
    font_color="black",
    font_weight="bold",
)

ax2.set_title(
    "2. Directed Transitive Graph\n(X → Y) and (Y → Z) ⟹ (X → Z) Shortcut Edge",
    fontsize=11,
    fontweight="bold",
    pad=15,
)
ax2.set_xlim(-1.6, 1.6)
ax2.set_ylim(-1.3, 1.3)
ax2.axis("off")

plt.tight_layout()
plt.subplots_adjust(bottom=0.28)
plt.show()
```
